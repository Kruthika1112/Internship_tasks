```python
# Q27
a = [10, 20]
b = a
a += [30]
print(b)
# the += operator modifies the list in-place rather than creating a new one, so both `a` and `b` still point to the same updated object.
```

```python
# Q35
import copy

a = [[1, 2], [3, 4]]
b = copy.copy(a)
b[0][1] = 99
print(a)
# copy.copy() creates a shallow copy, meaning the outer list is new but the inner nested lists are still references to the original objects
```

```python
# Q36
import copy

a = [[1, 2], [3, 4]]
b = copy.deepcopy(a)
b[0][1] = 99
print(a)
# opy.deepcopy() recursively copies all nested objects, making b completely independent, so changes to b have no effect on a
```

```python
# Q37
import copy

a = [[10], [20]]
b = a.copy()
a[1].append(30)
print(b)
# a.copy() creates a shallow copy, meaning the inner nested lists are shared references between a and b
```

```python
# Q38
import copy

a = [[10], [20]]
b = copy.deepcopy(a)
a[1].append(30)
print(b)
# because copy.deepcopy() creates completely independent nested objects, so modifying an inner list in a leaves b unaffected
```

```python
# Q39
import copy

a = [[1, 2], [3, 4]]
b = copy.copy(a)
print(a is b)
print(a[0] is b[0])
# copy.copy() creates a new outer list object, but copies references to the exact same inner nested lists
```

```python
# Q40
a = [[]] * 3
a[1].append(5)
print(a)
# because the * 3 syntax replicates the reference to the exact same inner list object three times, meaning modifying one modifies all of them
```

```python
# Q44
def test(*args):
    print(args)
test(1, 2)
# because the *args syntax captures all positional arguments passed to the function and packs them into a single tuple
```

```python
# Q45
def test(*args):
    print(type(args))

test("a", "b", "c")
# *args syntax always packs arbitrary positional arguments
```

```python
# Q46
def test(**kwargs):
    print(kwargs)

test(name="Ram", age=25)
# the **kwargs syntax captures all named keyword arguments passed to the function and packs them into a dictionary
```

```python
# Q47
def test(**kwargs):
    print(type(kwargs))

test(a=10, b=20)
# **kwargs syntax always packs arbitrary keyword arguments
```

```python
# Q48
def test():
    print(10)

x = test()
print(x)
# 10 and None (because the function runs and prints 10, but since it has no explicit return statement, it implicitly returns None, which is then assigned to x and printed)
```

```python
# Q49
def test():
    return 10

x = test()
print(x)
# the function explicitly passes the value 10 back via the return statement, which is then assigned to x and printed
```
